```python
# Load each file with pandas.read_csv(...)

import pandas as pd

# Load the datasets assuming they reside in your project directory
df_settlement = pd.read_csv('settlement_monitoring.csv')
df_groundwater = pd.read_csv('groundwater.csv')
df_inclinometer = pd.read_csv('inclinometer.csv')
df_tbm = pd.read_csv('TBM_data.csv')

# Verify successful ingestion by reviewing the shape (rows, columns) of each dataset
print(f"Settlement Records: {df_settlement.shape}")
print(f"Groundwater Records: {df_groundwater.shape}")
print(f"Inclinometer Records: {df_inclinometer.shape}")
print(f"TBM Operation Records: {df_tbm.shape}")
```

```python
# Convert the date strings to true datetime objects across all dataframes
df_settlement['date'] = pd.to_datetime(df_settlement['date'])
df_groundwater['date'] = pd.to_datetime(df_groundwater['date'])
df_inclinometer['date'] = pd.to_datetime(df_inclinometer['date'])
df_tbm['date'] = pd.to_datetime(df_tbm['date'])

# Verify that the data type has successfully updated to datetime64[ns]
print("Verified Date Column Data Types:")
print(f"Settlement:   {df_settlement['date'].dtype}")
print(f"Groundwater:  {df_groundwater['date'].dtype}")
print(f"Inclinometer: {df_inclinometer['date'].dtype}")
print(f"TBM Progress: {df_tbm['date'].dtype}")
```

```python
# - Plot settlement_mm against date for one point_id.
import matplotlib.pyplot as plt
import seaborn as sns

# Configure our layout style
sns.set_theme(style="whitegrid")
plt.figure(figsize=(11, 5))

# 1. Isolate and chronologically arrange data for point SP-001
df_point1 = df_settlement[df_settlement['point_id'] == 'SP-001'].sort_values('date')

# 2. Draw the time-series trajectory
plt.plot(
    df_point1['date'], 
    df_point1['settlement_mm'], 
    marker='o', 
    markersize=3, 
    linestyle='-', 
    color='#1f77b4', 
    label='SP-001 (Station ST-100)'
)

# 3. Geotechnical formatting rule: Invert the Y-axis so ground subsidence moves downward
plt.gca().invert_yaxis()

# 4. Finalize text layout
plt.title('Synthetic Ground Settlement Over Time — Instrument SP-001', fontsize=13, fontweight='bold', pad=12)
plt.xlabel('Date', fontsize=11)
plt.ylabel('Settlement (mm) [Inverted Axis]', fontsize=11)
plt.xticks(rotation=30)
plt.legend(loc='lower left')

plt.tight_layout()
plt.show()

```

```python
import datetime  # Added missing import for datetime.date.toordinal
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ==========================================
# 0. LOAD YOUR DATA (Choose Option A or B)
# ==========================================

# OPTION A: Uncomment the line below to load your actual CSV file
# df_settlement = pd.read_csv('your_file.csv')

# OPTION B: Creating synthetic data so this script runs immediately out-of-the-box
if 'df_settlement' not in locals() and 'df_settlement' not in globals():
    dates = pd.date_range(start='2026-01-01', periods=100, freq='D')
    # Generate mock settlement data trending downward (positive values = subsidence)
    settlement_values = np.linspace(0, 15, 100) + np.random.normal(0, 1.5, 100)
    
    df_settlement = pd.DataFrame({
        'date': dates,
        'point_id': ['SP-001'] * 100,
        'settlement_mm': settlement_values
    })

# Ensure the 'date' column is in a proper datetime format
df_settlement['date'] = pd.to_datetime(df_settlement['date'])


# ==========================================
# 1. ISOLATE AND SORT DATA
# ==========================================
sns.set_theme(style="whitegrid")

# Isolate and chronologically arrange data for point SP-001
df_point1 = df_settlement[df_settlement['point_id'] == 'SP-001'].sort_values('date')


# ==========================================
# CHART 1: Histogram (Distribution of Settlement)
# ==========================================
plt.figure(figsize=(10, 5))

# Plot histogram with a Kernel Density Estimate (KDE) curve
sns.histplot(
    data=df_point1, 
    x='settlement_mm', 
    bins=15, 
    kde=True, 
    color='#2ca02c', 
    edgecolor='white'
)

plt.title('Distribution of Settlement Values — Instrument SP-001', fontsize=13, fontweight='bold', pad=12)
plt.xlabel('Settlement (mm)', fontsize=11)
plt.ylabel('Frequency (Count)', fontsize=11)

plt.tight_layout()
plt.show()


# ==========================================
# CHART 2: Line Chart with Trendline
# ==========================================
plt.figure(figsize=(11, 5))

# Plot the raw time-series trajectory
plt.plot(
    df_point1['date'], 
    df_point1['settlement_mm'], 
    marker='o', 
    markersize=3, 
    linestyle='-', 
    color='#1f77b4', 
    label='Observed SP-001'
)

# Calculate a linear trendline (using ordinal dates for numeric regression)
x_numeric = df_point1['date'].map(datetime.date.toordinal)
slope, intercept = np.polyfit(x_numeric, df_point1['settlement_mm'], 1)
trendline_y = slope * x_numeric + intercept

# Plot the dashed trendline
plt.plot(
    df_point1['date'], 
    trendline_y, 
    linestyle='--', 
    color='#d62728', 
    linewidth=2, 
    label=f'Trendline (Slope: {slope:.4f} mm/day)'
)

# Geotechnical formatting rule: Invert the Y-axis so subsidence moves downward
plt.gca().invert_yaxis()

# Finalize text layout
plt.title('Ground Settlement Over Time with Trendline — Instrument SP-001', fontsize=13, fontweight='bold', pad=12)
plt.xlabel('Date', fontsize=11)
plt.ylabel('Settlement (mm) [Inverted Axis]', fontsize=11)
plt.xticks(rotation=30)
plt.legend(loc='lower left')

plt.tight_layout()
plt.show()
```

```python
# Group by station and calculate maximum settlement.
import pandas as pd

## 1. Group the data by structural station and find the maximum reading
df_max_settlement = df_settlement.groupby('station')['settlement_mm'].max().reset_index()

# 2. Rename the column for clear reporting
df_max_settlement = df_max_settlement.rename(columns={'settlement_mm': 'max_settlement_mm'})

# 3. Sort chronologically/hierarchically by severity (highest settlement first)
df_max_settlement = df_max_settlement.sort_values(by='max_settlement_mm', ascending=False).reset_index(drop=True)

# 4. View the results profile
print("Calculated Peak Settlement per Station:")
print(df_max_settlement)
```

```python
#- Join settlement and groundwater on date + 
# station for exploratory analysis.

# 1. Execute the dual-key merge
df_merged = pd.merge(
    df_settlement, 
    df_groundwater, 
    on=['date', 'station'], 
    how='inner'
)

# 2. Drop the separate 'data_type' string tags to avoid name duplication columns
df_merged = df_merged.drop(columns=['data_type_x', 'data_type_y'], errors='ignore')

# 3. Sort chronologically by station to keep the data organized
df_merged = df_merged.sort_values(by=['station', 'date']).reset_index(drop=True)

# 4. View a snapshot of your new unified exploration pipeline
print("Exploratory Data Structure Matrix Profile:")
print(df_merged.head())
print(f"\nTotal unified rows matched: {len(df_merged)}")
```
